```python
import re
from dataclasses import dataclass


@dataclass
class ATBShop:
    id: int
    short_name: str
    lat: float
    lon: float
    address: str
    worktime: str


_PREFIX_RE = re.compile(
    r'^(?:вул\.|бульв\.|просп\.|пров\.|шосе|Шосе|Кільцева дорога,?|Русанівська Набережна,?)\s*'
)
_OPTION_RE = re.compile(
    r"<option value='(\d+)' worktime='([^']+)' city='\d+'>([^<]+)</option>"
)


def parse_atb_shops(data: dict) -> list[ATBShop]:
    coord_map: dict[int, tuple[float, float]] = {
        item["id"]: (float(item["lat"]), float(item["lng"]))
        for item in data["coordinates"]
    }

    shops: list[ATBShop] = []
    for m in _OPTION_RE.finditer(data["optselect"]):
        shop_id = int(m.group(1))
        worktime = m.group(2)
        address = m.group(3)
        lat, lon = coord_map[shop_id]
        short_name = _PREFIX_RE.sub("", address).strip()

        shops.append(ATBShop(
            id=shop_id,
            short_name=short_name,
            lat=lat,
            lon=lon,
            address=address,
            worktime=worktime,
        ))

    return shops
parse_atb_shops(response.json())
```

```python
# re
import re

_OPTION_RE = re.compile(
    r"<option value='(\d+)' worktime='([^']+)' city='\d+'>([^<]+)</option>"
)

def parse_with_re(optselect: str) -> list[tuple]:
    return [
        (int(m.group(1)), m.group(2), m.group(3))
        for m in _OPTION_RE.finditer(optselect)
    ]

# lxml
from lxml import html

def parse_with_lxml(optselect: str) -> list[tuple]:
    tree = html.fromstring(f"<select>{optselect}</select>")
    return [
        (int(el.get("value")), el.get("worktime"), el.text_content())
        for el in tree.xpath("//option[@value != '']")
    ]

from selectolax.lexbor import LexborHTMLParser

def parse_with_lexbor(optselect: str) -> list[tuple]:
    tree = LexborHTMLParser(f"<select>{optselect}</select>")
    return [
        (int(el.attrs["value"]), el.attrs["worktime"], el.text())
        for el in tree.css("option[value]")
        if el.attrs["value"]
    ]

# benchmark
import timeit

data = response.json()["optselect"]  # ← ось де помилка

re_time = timeit.timeit(lambda: parse_with_re(data), number=10_000)
lxml_time = timeit.timeit(lambda: parse_with_lxml(data), number=10_000)
selectolax_time = timeit.timeit(lambda: parse_with_lexbor(data), number=10_000)
print(f"re:   {re_time:.3f}s")
print(f"lxml: {lxml_time:.3f}s")
print(f"selectolax: {selectolax_time:.3f}s")
```

```python
from src.models.products import ATBProduct


def parse_atb_product(item: dict) -> ATBProduct:
    old_price = item.get("oldprice")
    return ATBProduct(
        id=item["id"],
        name=item["name"],
        price=old_price if old_price else item["price"],
        in_stock=item["is_presence"],
        url=item["url"],
        image_url=item.get("picture"),
        special_price=item["price"] if old_price else None,
    )


def parse_atb_response(data: dict) -> list[ATBProduct]:
    products = []
    for group in data["results"]["item_groups"]:
        for item_list in group["items"]:
            for item in item_list:
                products.append(parse_atb_product(item))
    return products

parse_atb_response(response.json())
```
